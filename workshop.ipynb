## Graph object

A simple graph: $\mathcal{G} = (\mathcal{V}, \mathcal{E})$.

$\mathcal{G}$ - is the graph.

$\mathcal{V}$ - is a set (list) of nodes (V for vertex).

$\mathcal{E}$ - is a set (list) of edges.

## Brain Networks

::: {.incremental}
- fMRI: we have timeseries from brain areas and infer nodes. 
- EEG/MEG: we have timeseries from sensors or source areas and infer nodes, and then infer edges. 
- DTI: we have anatomical connections between brain areas.
:::

## Edge inference: 

A non-exhaustive list of methods to infer edges: 

<small>Pearson correlations, Spearman correlations, Bayesian estimates, partial (directed) correlations, Ridge/GLasso Regression, Granger Causality, dynamic causal modelling, phase synchrony, transfer entropy, coherence, coherency, phase locking value, weighted phase lag, spike-field coherence, spike correlations, canonical-correlation analysis, Hamming distance.</small>

::: {.fragment}
So which should you choose?

# Demonstration

## Network as object and network as array

```python
#| echo: true
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
import bct
```

## Adding a Simple Network

```python
#| echo: true
# Create an empty graph
G = nx.Graph()

# Add nodes
G.add_nodes_from([1, 2, 3, 4, 5])

# Add edges
G.add_edges_from([(1, 2), (2, 3), (3, 4), (4, 5), (5, 1)])
```

## Understanding the Network

```python
#| echo: true
# Get the number of nodes
print(G.number_of_nodes())

# Get the number of edges
print(G.number_of_edges())
```

##

```python
#| echo: true
# Get the list of nodes
print(G.nodes())

# Get the list of edges
print(G.edges())
```

## Checking Node and Edge Attributes

Currently, there are no attributes assigned to the nodes or edges.

```python
#| echo: true
# Retrieve node attributes (empty by default)
G.nodes[1]
```

```python
#| echo: true
# Retrieve edge attributes (empty by default)
G[1][2]  # or G.edges[1, 2]
```

## Quantifying the network

There are a lot of measures. 

```python
#| echo: true
nx.efficiency_measures.global_efficiency(G)
```

## Visualizing the Network

```python
#| echo: true
# Visualize the network
nx.draw(G, with_labels=True)
plt.show()
```

##  Adding Weights to the Edges

More commonly, weights are predefined in datasets. 
Consider the following adjacency matrix:

|       | Ashley | Blake | Casey | Dylan | Elliot | Finley |
|-------|--------|--------|--------|--------|--------|--------|
| Ashley | 0      | 2      | 0      | 0      | 2      | 1      |
| Blake  | 2      | 0      | 4      | 0      | 0      | 0      |
| Casey  | 0      | 4      | 0      | 3      | 0      | 0      |
| Dylan  | 0      | 0      | 3      | 0      | 1      | 5      |
| Elliot | 2      | 0      | 0      | 1      | 0      | 0      |
| Finley | 1      | 0      | 0      | 5      | 0      | 0      |

## 

```python
#| echo: true
import pandas as pd

# Create a dataframe above as an adjacency matrix
df = pd.DataFrame(data={
    'Ashley': [0, 2, 0, 0, 2, 1],
    'Blake': [2, 0, 4, 0, 0, 0],
    'Casey': [0, 4, 0, 3, 0, 0],
    'Dylan': [0, 0, 3, 0, 1, 5],
    'Elliot': [2, 0, 0, 1, 0, 0],
    'Finley': [1, 0, 0, 5, 0, 0]
}, index=['Ashley', 'Blake', 'Casey', 'Dylan', 'Elliot', 'Finley'])

G = nx.from_pandas_adjacency(df)
```

## 
To retrieve node information using names:

```python
#| echo: true
print(G.nodes['Ashley'])  # <- still empty by default
print(G.edges['Ashley', 'Blake'])
print(G['Ashley']['Finley'])
```

## Different Layouts (circle)


```python
#| echo: true
plt.figure(figsize=(6, 4))
# Circular layout
nx.draw(G, with_labels=True, pos=nx.circular_layout(G))
plt.title('Circular Layout')
```

## Different Layouts (Random)


```python
#| echo: true
plt.figure(figsize=(6, 4))
# Random layout, specify seed for reproducing the same layout
pos = nx.random_layout(G, seed=42)
nx.draw(G, with_labels=True, pos=pos)
plt.title('Random Layout')
```

## Different Layouts (Spring)

```python
#| echo: true
# Spring layout
plt.figure(figsize=(6, 4))
nx.draw(G, with_labels=True, pos=nx.spring_layout(G))
plt.title('Spring Layout')
plt.show()
```

## Summary / Cheat Sheet

| <small>Concept</small> | <small>`networkx` Function</small> |
------------------------|----------------------|
| <small>Create network</small>        | <small>`nx.Graph()`, `nx.from_pandas_adjacency()`, `nx.from_numpy_matrix()`</small> |
| <small>Network properties</small>      | <small>`G.number_of_nodes()`, `G.number_of_edges()`, `G.nodes()`, `G.edges()`</small> |
| <small>Visualizing network</small>     | <small>`nx.draw(G, with_labels=True, layout=...)`</small> |
| <small>Add nodes/edges</small>         | <small>`G.add_node()`, `G.add_nodes_from()`, `G.add_edge()`, `G.add_edges_from()`</small> |
| <small>Name nodes</small>              | <small>`nx.set_node_attributes(G, {1: "A", 2: "B"}, "name")`, `nx.get_edge_attributes("name")`</small> |
| <small>Edge weights</small>            | <small>`nx.set_edge_attributes(G, {(1, 2): 5, (2, 3): 8}, "weight")`, `nx.get_edge_attributes("weight")`</small> |
| <small>Layouts</small>                 | <small>`nx.circular_layout(G)`, `nx.random_layout(G)`, `nx.spring_layout(G)`, etc.</small> |
